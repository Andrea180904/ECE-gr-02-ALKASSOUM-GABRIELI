## Chaîned function with all transformations

```python
def text_processing_pipeline(rdd, stop_words, punctuation):

    import re

    def clean_word(word):
        # Remove punctuation from words
        word = re.sub(f'[{re.escape(punctuation)}]', '', word)
        return word

    # Split text into words, clean, filter stop words, count occurrences, sort by frequency
    return (rdd.flatMap(lambda lines: lines.split(' '))
            .map(lambda word: clean_word(word.lower()))
            .filter(lambda word: word.strip() != '' and word not in stop_words)
            .map(lambda word: (word, 1))
            .reduceByKey(lambda a, b: a + b)
            .sortBy(lambda x: x[1], ascending=False))


# Define common English stop words and punctuation to remove
stop_words = {'a', 'an', 'and', 'are', 'as', 'at', 'be', 'by', 'for',
              'from', 'has', 'he', 'in', 'is', 'it', 'its', 'of', 'on',
              'or', 'that', 'the', 'to', 'was', 'will', 'with'}
punctuation = ".,!?;:'-\"()[]{}—–"

# Run pipeline and display top 20 words
result = text_processing_pipeline(rdd, stop_words, punctuation)
result.take(20)
```

## 3. Function timing.

- write a simple python timer function for seeing how quickly your rdd runs as written. change the order of the steps in order to make the rdd run as optimally as possible

```python
import time

def time_rdd_operation(rdd_func, operation_name):
    """Timer for RDD operations."""
    start = time.time()
    result = rdd_func()
    # Force evaluation with collect()
    _ = result.collect()
    elapsed = time.time() - start
    print(f"{operation_name}: {elapsed:.4f} seconds")
    return elapsed

# Inefficient: aggregates words before filtering stop words
def original_order():
    stop_words = {'a', 'an', 'and', 'are', 'as', 'at', 'be', 'by', 'for',
                  'from', 'has', 'he', 'in', 'is', 'it', 'its', 'of', 'on',
                  'or', 'that', 'the', 'to', 'was', 'will', 'with'}
    return (rdd.flatMap(lambda lines: lines.split(' '))
            .map(lambda word: word.lower())
            .map(lambda word: (word, 1))
            .reduceByKey(lambda a, b: a + b)
            .filter(lambda x: x[0] not in stop_words))

# Optimized: filters stop words before expensive reduceByKey operation
def optimized_order():
    stop_words = {'a', 'an', 'and', 'are', 'as', 'at', 'be', 'by', 'for',
                  'from', 'has', 'he', 'in', 'is', 'it', 'its', 'of', 'on',
                  'or', 'that', 'the', 'to', 'was', 'will', 'with'}
    return (rdd.flatMap(lambda lines: lines.split(' '))
            .map(lambda word: word.lower())
            .filter(lambda word: word not in stop_words)
            .map(lambda word: (word, 1))
            .reduceByKey(lambda a, b: a + b))

# Compare performance of both approaches
t1 = time_rdd_operation(original_order, "Original order")
t2 = time_rdd_operation(optimized_order, "Optimized order")
print(f"\nImprovement: {((t1-t2)/t1)*100:.2f}% faster")
```

- perform eda on the original french version of the [book](https://www.gutenberg.org/ebooks/46541.txt.utf-8) and compare the two

```python
# Download French version
!wget -nc -O tour_du_monde_80_jours.txt https://www.gutenberg.org/ebooks/46541.txt.utf-8
```

```python
# Load and process both versions
rdd_en = sc.textFile('around_the_world_in_80_days.txt')
rdd_fr = sc.textFile('tour_du_monde_80_jours.txt')

stop_words_en = {'a', 'an', 'and', 'are', 'as', 'at', 'be', 'by', 'for',
                 'from', 'has', 'he', 'in', 'is', 'it', 'its', 'of', 'on',
                 'or', 'that', 'the', 'to', 'was', 'will', 'with'}
stop_words_fr = {'à', 'and', 'au', 'aux', 'avec', 'ce', 'ces', 'dans', 'de',
                 'des', 'du', 'elle', 'en', 'et', 'eux', 'for', 'il', 'je',
                 'la', 'le', 'les', 'leur', 'lui', 'ma', 'me', 'même', 'mes',
                 'moi', 'mon', 'ne', 'nos', 'nous', 'on', 'ou', 'par', 'pas',
                 'plus', 'pour', 'qu', 'que', 'qui', 'sa', 'se', 'ses', 'si',
                 'son', 'sur', 'ta', 'te', 'tes', 'toi', 'ton', 'tu', 'un',
                 'une', 'vos', 'vous', 'y', 'ça', 'était', 'étaient', 'être'}

punctuation = ".,!?;:'-\"()[]{}—–"
import re

def clean_word(word):
    word = re.sub(f'[{re.escape(punctuation)}]', '', word)
    return word

# Process English version
word_count_en = (rdd_en.flatMap(lambda lines: lines.split(' '))
                  .map(lambda word: clean_word(word.lower()))
                  .filter(lambda word: word.strip() != '' and word not in stop_words_en)
                  .map(lambda word: (word, 1))
                  .reduceByKey(lambda a, b: a + b)
                  .sortBy(lambda x: x[1], ascending=False))

# Process French version
word_count_fr = (rdd_fr.flatMap(lambda lines: lines.split(' '))
                  .map(lambda word: clean_word(word.lower()))
                  .filter(lambda word: word.strip() != '' and word not in stop_words_fr)
                  .map(lambda word: (word, 1))
                  .reduceByKey(lambda a, b: a + b)
                  .sortBy(lambda x: x[1], ascending=False))

print("Top 15 words in English version:")
for word, count in word_count_en.take(15):
    print(f"  {word}: {count}")

print("\nTop 15 words in French version:")
for word, count in word_count_fr.take(15):
    print(f"  {word}: {count}")
```

```python
# Comparison stats
total_en = word_count_en.map(lambda x: x[1]).sum()
total_fr = word_count_fr.map(lambda x: x[1]).sum()
unique_words_en = word_count_en.count()
unique_words_fr = word_count_fr.count()

print(f"English version: {total_en} words, {unique_words_en} unique words")
print(f"French version: {total_fr} words, {unique_words_fr} unique words")
print(f"\nAverage word frequency:")
print(f"  English: {total_en/unique_words_en:.2f}")
print(f"  French: {total_fr/unique_words_fr:.2f}")
```
